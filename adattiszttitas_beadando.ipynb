```python
#2. Az Egységár és a Érték oszlopot alakítsa egész típusúvá.
def masodik_feladat(oszlop1: str, oszlop2: str, df: pd.DataFrame) -> pd.DataFrame:

    try:

        if not isinstance(oszlop1, str) or not isinstance(oszlop2, str):
            raise TypeError("Oszlopnév szöveg kell legyen!")
        
        if oszlop1 not in df.columns or oszlop2 not in df.columns:
            raise KeyError("Oszlop nem létezik!")

        df_clean = df_2.copy()

        df_clean[oszlop1] = df_clean[oszlop1].fillna('0')
        df_clean[oszlop2] = df_clean[oszlop2].fillna('0')

        
        df_clean[oszlop1] = df_clean[oszlop1].astype(str)
        df_clean[oszlop1] = df_clean[oszlop1].str.replace(',', '.')
        df_clean[oszlop1] = df_clean[oszlop1].str.replace(r'Ft', '', regex=True)
        df_clean[oszlop1] = df_clean[oszlop1].str.replace(r'[^\d.]', '', regex=True)
        
        df_clean[oszlop2] = df_clean[oszlop2].astype(str)  # ← KULCS! (hiányzott!)
        df_clean[oszlop2] = df_clean[oszlop2].str.replace(',', '.')
        df_clean[oszlop2] = df_clean[oszlop2].str.replace(r'[^\d.]', '', regex=True)
        

        
        df_clean[oszlop1] = pd.to_numeric(df_clean[oszlop1], errors='coerce').astype('Int64')
        df_clean[oszlop2] = pd.to_numeric(df_clean[oszlop2], errors='coerce').astype('Int64')
        
        
        return df_clean
    
    except (TypeError, KeyError) as e:
        print(f" Hiba: {e}")
        return None
    except Exception as e:
        print(f" Ismeretlen hiba: {type(e).__name__}: {e}")
        return None

df_clean = masodik_feladat("Egységár", "Érték", df_2)
df_2=df_clean
```
